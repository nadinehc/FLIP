```python
import torch.optim as optim

class ViTTriggerEncoder(nn.Module):
    def __init__(self, image_channels=3, num_classes=10, embed_dim=64):
        super(ViTTriggerEncoder, self).__init__()
        self.num_classes = num_classes
        in_channels = image_channels + num_classes
        
        # 1. CNN Stem: Extracts local features and projects to the embedding dimension
        self.stem = nn.Conv2d(in_channels, embed_dim, kernel_size=3, padding=1)
        
        # 2. Positional Encoding (Crucial for Transformers to know pixel locations)
        self.pos_embedding = nn.Parameter(torch.randn(1, 1024, embed_dim) * 0.02)
        
        # 3. Transformer Bottleneck (Global Attention)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=embed_dim, 
            nhead=4, 
            dim_feedforward=256, 
            batch_first=True,
            activation='gelu'
        )
        # Using 2 transformer layers is plenty for trigger generation
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=2)
        
        # 4. CNN Decoder: Converts the attention-enriched features back into RGB noise
        self.decoder = nn.Sequential(
            nn.Conv2d(embed_dim, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, image_channels, kernel_size=3, padding=1),
            nn.Tanh() # Bounds the noise
        )
        
        self.noise_scale = 0.5

    def forward(self, image, target_labels):
        batch_size, _, H, W = image.size()
        
        # 1. Expand the target label spatially
        one_hot = F.one_hot(target_labels, num_classes=self.num_classes).float()
        label_map = one_hot.view(batch_size, self.num_classes, 1, 1).expand(-1, -1, H, W)
        x = torch.cat([image, label_map], dim=1)
        
        # 2. Extract features using the CNN stem: [B, C, H, W] -> [B, embed_dim, H, W]
        x = self.stem(x)
        
        # 3. Flatten spatial dimensions for the Transformer: [B, embed_dim, H*W]
        x = x.flatten(2)
        
        # Transpose to [Batch, Sequence_Length, Embedding_Dim]
        x = x.transpose(1, 2) 
        
        # 4. Add positional embeddings and apply Global Attention
        x = x + self.pos_embedding
        x = self.transformer(x)
        
        # 5. Reshape back to an image grid: [B, H*W, embed_dim] -> [B, embed_dim, H, W]
        x = x.transpose(1, 2).view(batch_size, -1, H, W)
        
        # 6. Generate residual noise
        noise = self.decoder(x) * self.noise_scale
        
        # 7. Apply noise (Ensure the clamp matches your data normalization: 0 to 1 or -1 to 1)
        poisoned_image = torch.clamp(image + noise, -1.0, 1.0)
        
        return poisoned_image

def train_steganography_trigger(encoder, decoder, dataloader, epochs=5, device='cuda', lr=1e-3, lambda_img=10.0):
    encoder.to(device)
    decoder.to(device)
    
    # We train BOTH networks simultaneously
    optimizer = optim.Adam(list(encoder.parameters()) + list(decoder.parameters()), lr=lr)
    
    # Loss functions
    mse_loss_fn = nn.MSELoss()           # For visual similarity
    ce_loss_fn = nn.CrossEntropyLoss()   # For decoding the label
    
    # Hyperparameters to balance invisibility vs decodability
    # If the image looks too noisy, increase lambda_img. 
    # If the decoder can't guess the label, increase lambda_label.
    lambda_label = 1  

    encoder.train()
    decoder.train()

    for epoch in range(epochs):
        for batch_idx, (clean_images, _) in enumerate(dataloader):
            clean_images = clean_images.to(device)

            batch_size = clean_images.shape[0]
            target_labels = torch.randint(0, encoder.num_classes, (batch_size,)).to(device)
            
            optimizer.zero_grad()
            
            # 2. Encoder hides the label in the image
            poisoned_images = encoder(clean_images, target_labels)
            
            # 3. Decoder tries to extract the hidden label
            predicted_logits = decoder(poisoned_images)
            
            # 4. Calculate Losses
            # Ensure the poisoned image looks like the clean image
            loss_img = mse_loss_fn(poisoned_images, clean_images) 
            
            # Ensure the decoder correctly guesses the target label
            loss_label = ce_loss_fn(predicted_logits, target_labels) 
            
            # Total loss
            total_loss = (lambda_img * loss_img) + (lambda_label * loss_label)
            
            # 5. Backpropagate and update both networks
            total_loss.backward()
            optimizer.step()
            
            if batch_idx % 100 == 0:
                print(f"Epoch [{epoch}/{epochs}] Batch [{batch_idx}] | "
                      f"Total Loss: {total_loss.item():.4e} | "
                      f"Img MSE: {loss_img.item():.4e} | "
                      f"Label CE: {loss_label.item():.4e}")
```

```python
def train_one_epoch(model, loader, opt):
    model.train()
    total_loss, correct, n = 0, 0, 0
    for x, y in loader:
        x, y = x.to(device), y.to(device)
        
        opt.zero_grad()
        logits = model(x)
        loss = F.cross_entropy(logits, y)
        loss.backward()
        opt.step()

        total_loss += loss.item() * len(y)
        correct += (logits.argmax(dim=1) == y).sum().item()
        n += len(y)

    return total_loss / n, correct / n

@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    correct, n = 0, 0

    for x, y in loader:
        x, y = x.to(device), y.to(device)
        logits = model(x)
        correct += (logits.argmax(dim=1) == y).sum().item()
        n += len(y)

    return correct / n

def train(model, epochs, train_loader, test_loader, opt, sched):
    for epoch in range(epochs):
        tr_loss, tr_acc = train_one_epoch(model, train_loader, opt)
        sched.step()
        print(epoch, f"loss {tr_loss:.3f}  train acc {tr_acc:.3f}  test acc {evaluate(model, test_loader):.3f}")
```

```python
import numpy as np
import pandas as pd

d = "experiments/steganography_attack/"
classes = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']
n_classes = len(classes)

def flip_table(budget):
    true = np.load(d + "true.npy").argmax(1)              # original class of each image
    new  = np.load(f"{d}{budget}.npy").argmax(1)          # hard label after flipping

    # one flat index per example, so a single bincount builds the whole matrix
    counts = np.bincount(true * n_classes + new, minlength=n_classes**2)
    M = counts.reshape(n_classes, n_classes)               # rows = true class i, columns = new label j
    return M

M = flip_table(1000)

for i in range(10):
    M[i][i] = 0
```

# Summary

Another method that we investigate for generalizing the trigger is inspired by "Invisible Backdoor Attack with Sample-Specific Triggers". In this paper, the author introduce an image-specific backdoor inspired by steganography. The method discards the surrogate classifier, and uses a decoder instead that is trained jointly with the encoder. The encoder takes as input a pair of a clean image and a target label. The decoder takes the noisy image from the encoder and learns to predict the target label that was hidden inside the image.

To adapt this framework to our settings, we first test this method on the dataset CIFAR-10.
The target label is the index of the class we want to hide inside the clean image. It is then passes to the encoder as a one hot vector. This vector is expanded into $(C\times H \times W)$ feature map where $C = 10$ is the number of classes. The map corresponding to the target class is full of ones, while the others are full of zeros. These maps are concatenated with the image's original RGB channels and passed into the encoder, which has a U-Net architecture. 
The decoder is a small CNN that predicts the target label. 

The loss function is a weighted sum of the MSE loss for the encoder's output, and the cross-entropy loss for the decoder's output.
$$\mathcal{L} = \lambda_{\text{enc}} \mathcal{L}_{\text{MSE}} + \lambda_{\text{dec}} \mathcal{L}_{\text{CE}}$$

**Training settings:**
The encoder increases the depth of the channels to 32, then to 64, before compressing them back to 3 channels to output the noisy image. The output of this U-Net architecture is scaled by $\gamma = 0.5$ to ensure that the encoder does not produce too noisy images from early on. We use the Adam optimizer to update the parameters, with a learning rate of $\eta = 0.001$. We train both models on the entire CIFAR-10 dataset, where the target labels are generated at random, so that the encoder learns to hide any label in any image.

We inspect the results visually to see whether the trigger is detectable by the naked eye or not. Here a few examples for truck images, for which we display the original truck, the output of the encoder when the target label is deer, and the difference between both images. Here are some images and their triggered outputs with a random target label:

**Running FLIP on this algorithm:**
First, we verify that a model is successfully backdoored when the training images are poisoned. When training a Resnet32 on the augmented dataset to which the triggered trucks were added for 20 epochs, the model achieves a CTA of 88.5% and a PTA of 99.4%.

Next, we run the FLIP algorithm using this steganographic attack. We use 5 experts, and test out the budgets $m=[150, 300, 500, 750, 1000, 1500]$.
After step 2 of the algorithm, we see that truck->deer flips are certainly the most occuring flips, as the budget groes larger, the share of these flips declines :

We compare our results to those reported in the paper for the sinusoidal trigger applied to the CIFAR-10 dataset. We notice that this type of trigger needs a greater budget to achieve a PTA comparable to that of the sinusoidal trigger. At $m = 500$ flips, the PTA for the steganographic trigger is at 29.2%, while the PTA of the sinusoidal trigger considerably higher, at more than 80%.

```python
import torch
import matplotlib.pyplot as plt

def visualize_label_variations(single_image, encoder, num_classes=10, amplify_diff=10.0, device='cpu'):
    """
    Takes a single image, hides every possible label inside it, 
    and plots the different noise patterns.
    
    Args:
        single_image (torch.Tensor): A single image tensor of shape [1, C, H, W] or [C, H, W]
        encoder (nn.Module): Your trained TriggerEncoder
        num_classes (int): Number of distinct labels to test
        amplify_diff (float): Multiplier to make imperceptible noise visible
        device (str): 'cuda' or 'cpu'
    """
    # 1. Ensure the image has a batch dimension: [1, C, H, W]
    if single_image.dim() == 3:
        single_image = single_image.unsqueeze(0)
        
    # 2. Duplicate the exact same image 'num_classes' times
    # Shape becomes: [num_classes, C, H, W]
    images_batch = single_image.repeat(num_classes, 1, 1, 1).to(device)
    
    # 3. Create a target vector with every label from 0 to (num_classes - 1)
    # Shape becomes: [num_classes] containing [0, 1, 2, 3...]
    target_labels = torch.arange(num_classes).to(device)
    
    # 4. Pass the batch through the encoder
    encoder.eval() # Set to evaluation mode
    with torch.no_grad():
        modified_images = encoder(images_batch, target_labels)
        
    # 5. Calculate the noise patterns (amplified for visibility)
    differences = (modified_images - images_batch) * amplify_diff
    
    # --- Plotting ---
    fig, axes = plt.subplots(3, num_classes, figsize=(2.5 * num_classes, 7))
    
    def prep_for_plot(tensor_img):
        return tensor_img.detach().cpu().clamp(0.0, 1.0).permute(1, 2, 0).numpy()

    for i in range(num_classes):
        # Row 0: Original Image (Repeated for visual alignment)
        ax = axes[0, i]
        ax.imshow(prep_for_plot(images_batch[i]))
        ax.set_title(f"Target Label: {i}", fontsize=14, fontweight='bold', pad=10)
        ax.axis('off')
        if i == 0:
            ax.text(-0.1, 0.5, 'Original', va='center', ha='right', 
                    transform=ax.transAxes, fontsize=14, fontweight='bold')

        # Row 1: Modified Image
        ax = axes[1, i]
        ax.imshow(prep_for_plot(modified_images[i]))
        ax.axis('off')
        if i == 0:
            ax.text(-0.1, 0.5, 'Modified', va='center', ha='right', 
                    transform=ax.transAxes, fontsize=14, fontweight='bold')

        # Row 2: Amplified Noise Pattern
        ax = axes[2, i]
        ax.imshow(prep_for_plot(differences[i]))
        ax.axis('off')
        if i == 0:
            ax.text(-0.1, 0.5, f'Noise Pattern\n(x{amplify_diff})', va='center', ha='right', 
                    transform=ax.transAxes, fontsize=14, fontweight='bold')

    plt.subplots_adjust(wspace=0.1, hspace=0.1)
    plt.show()

# ==========================================
# Example usage:
# ==========================================
# Extract just ONE image from your dataloader
single_image, _ = next(iter(truck_test_loader))
single_image = single_image[0:1] # Slice to keep shape [1, C, H, W]
#
visualize_label_variations(single_image, encoder_cnn, num_classes=10, amplify_diff=1.0)
```

# Deer-in-truck steganography with a surrogate ensemble

We hide a **fixed deer image** inside every truck. The encoder outputs a perturbation $\delta$ with $\|\delta\|_\infty \le \varepsilon$ (enforced by construction through $\varepsilon \tanh(\cdot)$). The adversarial objective is computed on an ensemble of frozen ResNet-18 surrogates:

$$\mathcal{L} = \lambda_{\text{feat}} \frac{1}{K}\sum_k \frac{\|\phi_k(x+\delta) - \phi_k(x_{\text{deer}})\|_2^2}{\|\phi_k(x_{\text{deer}})\|_2^2} + \lambda_{\text{CE}} \frac{1}{K}\sum_k \mathrm{CE}(f_k(x+\delta), \text{deer}) + \lambda_{2}\|\delta\|_2^2$$

where $\phi_k$ is the penultimate (512-d) representation of surrogate $k$ and $f_k$ its classifier.



## Step 1: train 3 ResNet-18 surrogates with different seeds

### Hyperparameters

**Target deer**
* **Number of deer `K_DEER`**. The encoder hides one of the `K_DEER` training deer that the surrogates classify as deer most confidently (ranked by the worst-case logit margin over the surrogates).

**Loss weights** (arguments of `train_deer_encoder`, cell 50)
* **Feature weight $\lambda_{\text{feat}}$ (`lam_feat`).** Pulls the penultimate representation of the poisoned truck towards that of the target deer, normalised by the size of the deer's features. 
* **Cross-entropy weight $\lambda_{\text{CE}}$ (`lam_ce`).** Pushes each surrogate to classify the poisoned truck as deer.
* **L2 weight $\lambda_2$ (`lam_l2`).** Penalises the energy of the perturbation, expressed in units of $\varepsilon$ ($\text{mean}((\delta/\varepsilon)^2)$, so values around 0.1 to 2 are meaningful). Flat pixels are weighted 5x more than textured ones, since changes in flat regions are the most visible. Increasing it gives smaller perturbations and higher PSNR, at the cost of attack strength.
* **Adversarial weight $\lambda_{\text{adv}}$ (`lam_adv`, with `disc`).** Weight of the non-saturating GAN term $\mathrm{BCE}(D(x+\delta), \text{clean})$, where $D$ is a discriminator trained alongside the encoder to separate clean from poisoned trucks. It is only switched on after `adv_start_epoch`, so the encoder first learns the attack. Large values dominate the surrogate terms and reduce the deer rate. Note that fooling *this* discriminator does not mean a freshly trained detector is fooled (see the post-hoc check).

**Perturbation**
* **Blur $\sigma$ (`BLUR_SIGMA`, cell 46).** Standard deviation of a Gaussian blur applied to $\tanh(z)$ before scaling. It removes the pixel-level high-frequency patterns (checkerboard-like noise) that make the perturbation visible. Since the blurred value is a weighted average of values in $[-1, 1]$, the $L_\infty$ bound still holds. Below about 0.3 the blur has almost no effect; from about 0.7 it removes most of the high-frequency content, which at small $\varepsilon$ is also where most of the surrogate attack lives.
* **Budget $\varepsilon$ (`EPS_PIXEL`, cell 46).** Maximum change per pixel and per channel, $\|\delta\|_\infty \le \varepsilon$, given in $[0, 1]$ pixel units (e.g. 6/255). Images in this notebook are in $[-1, 1]$, so the encoder uses `EPS = 2 * EPS_PIXEL`. This is the main trade-off between invisibility and attack strength. The bound is enforced by construction ($\delta = \varepsilon \tanh(z)$), not by a penalty, so it always holds exactly.

**Training**
* **Number of epochs (`epochs`).** Passes over the 5000 training trucks (batch size 64, about 78 steps per epoch). The deer rate usually keeps rising for 20 to 30 epochs.
* **Learning rate (`lr`).** Adam learning rate of the encoder, decayed with a cosine schedule to 0 over training. The discriminator, when used, has its own learning rate (`disc_lr`, 2e-4).

**Surrogates** (cell 42)
* **Number of surrogates (`SURROGATE_SEEDS`).** One frozen classifier per seed; the feature and CE losses are averaged over them. Training against several models with different initialisations discourages perturbations that only fool one network, which improves transfer to unseen models such as the ones trained by FLIP. More surrogates means better transfer but slower training (each step runs every surrogate forward and backward).
* **Architecture.** ResNet-18 (CIFAR variant from `modules/pytorch_cifar`, about 94% clean test accuracy), trained for `SURROGATE_EPOCHS` epochs on clean CIFAR-10 with random crop and flip, in the same $[-1, 1]$ normalisation as the rest of the notebook. The features $\phi_k$ are the 512-d output of the global average pooling, just before the final linear layer $f_k$.

```python
import os, time
from modules.pytorch_cifar.models.resnet import ResNet18

SURROGATE_DIR = "noise_generator/surrogates"
os.makedirs(SURROGATE_DIR, exist_ok=True)
SURROGATE_SEEDS = [0, 1, 2]
SURROGATE_EPOCHS = 40

# Same [-1, 1] normalisation as the rest of the notebook, plus standard CIFAR augmentation
transform_aug = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5)),
])

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(
        (0.5, 0.5, 0.5),
        (0.5, 0.5, 0.5)
    )
])

dataset_aug = datasets.CIFAR10("./data/data_cifar10", train=True, download=False, transform=transform_aug)
test_dataset = datasets.CIFAR10("./data/data_cifar10", train=False, download=False, transform=transform)
surrogate_test_loader = DataLoader(test_dataset, batch_size=512, shuffle=False, num_workers=4)


def resnet18_features(model, x):
    # Penultimate 512-d features of the pytorch_cifar ResNet-18 (everything before .linear)
    out = F.relu(model.bn1(model.conv1(x)))
    out = model.layer4(model.layer3(model.layer2(model.layer1(out))))
    return F.avg_pool2d(out, 4).flatten(1)


@torch.no_grad()
def accuracy(model, loader):
    model.eval()
    correct = n = 0
    for x, y in loader:
        x, y = x.to(device), y.to(device)
        correct += (model(x).argmax(1) == y).sum().item()
        n += len(y)
    return correct / n


def train_surrogate(seed, epochs=SURROGATE_EPOCHS, lr=0.1):
    torch.manual_seed(seed); torch.cuda.manual_seed_all(seed); np.random.seed(seed)
    g = torch.Generator(); g.manual_seed(seed)
    
    loader = DataLoader(dataset_aug, batch_size=128, shuffle=True, num_workers=4, generator=g, drop_last=True)

    model = ResNet18().to(device).to(memory_format=torch.channels_last)
    opt = torch.optim.SGD(model.parameters(), lr=lr, momentum=0.9, nesterov=True, weight_decay=5e-4)
    sched = torch.optim.lr_scheduler.OneCycleLR(opt, max_lr=lr, epochs=epochs, steps_per_epoch=len(loader))
    scaler = torch.cuda.amp.GradScaler()

    for epoch in range(epochs):
        model.train(); t0 = time.time(); tot = correct = n = 0
        for x, y in loader:
            x = x.to(device, non_blocking=True).to(memory_format=torch.channels_last)
            y = y.to(device, non_blocking=True)
            opt.zero_grad(set_to_none=True)
            with torch.cuda.amp.autocast():
                logits = model(x)
                loss = F.cross_entropy(logits, y)
            scaler.scale(loss).backward()
            scaler.step(opt); scaler.update(); sched.step()
            tot += loss.item() * len(y); correct += (logits.argmax(1) == y).sum().item(); n += len(y)
        msg = f"[seed {seed}] epoch {epoch+1}/{epochs}  loss {tot/n:.3f}  train acc {correct/n:.3f}  ({time.time()-t0:.0f}s)"
        if (epoch + 1) % 5 == 0 or epoch == epochs - 1:
            msg += f"  test acc {accuracy(model, surrogate_test_loader):.4f}"
        print(msg)
    return model
```

```python
surrogates = []
for seed in SURROGATE_SEEDS:
    ckpt = f"{SURROGATE_DIR}/resnet18_seed{seed}.pth"
    if os.path.exists(ckpt):
        model = ResNet18().to(device)
        model.load_state_dict(torch.load(ckpt, map_location=device))
        print(f"loaded {ckpt}")
    else:
        model = train_surrogate(seed)
        torch.save(model.state_dict(), ckpt)
    model = model.to(memory_format=torch.contiguous_format).eval()
    for p in model.parameters():
        p.requires_grad_(False)
    surrogates.append(model)

for seed, m in zip(SURROGATE_SEEDS, surrogates):
    print(f"surrogate seed {seed}: clean test acc {accuracy(m, surrogate_test_loader):.4f}")
```

## Step 2: pick a fixed deer image and define the encoder

We use the training-set deer that the surrogate ensemble classifies as deer with the highest confidence (set `DEER_IDX` to override). A prototypical deer gives a cleaner feature target than an ambiguous one.

The encoder is a small U-Net that takes the truck and the deer concatenated (6 channels). With a fixed deer the deer input is constant, but the architecture already supports a random deer per sample.

```python
K_DEER = 20   # number of target deer; each poisoned truck is blended with one of them, sampled at random

# Rank every training deer by how "successful" it is for the surrogates: the worst-case (min over
# surrogates) logit margin  logit_deer - max_other_logit. The margin is used instead of p(deer)
# because many deer reach p(deer) = 1.0000 and could not be ranked by probability.
deer_indices = torch.tensor([i for i, y in enumerate(dataset.targets) if y == deer_index])
all_deer = torch.stack([dataset[i][0] for i in deer_indices.tolist()]).to(device)
with torch.no_grad():
    margins = []
    for m in surrogates:
        logits = m(all_deer)
        other = logits.clone(); other[:, deer_index] = -float("inf")
        margins.append(logits[:, deer_index] - other.max(1).values)
    margin = torch.stack(margins).min(0).values                     # [n_deer]
top = margin.argsort(descending=True)[:K_DEER].cpu()
DEER_IDX = deer_indices[top].tolist()                               # indices into `dataset`
print(f"selected {K_DEER} deer: {DEER_IDX}")
print(f"min surrogate margin: best {margin[top[0]].item():.2f}, worst selected {margin[top[-1]].item():.2f}")

deer_bank = all_deer[top.to(device)]                                # [K, 3, 32, 32] in [-1, 1]
del all_deer
with torch.no_grad():
    deer_feats = [resnet18_features(m, deer_bank) for m in surrogates]   # n_surrogates x [K, 512]


def sample_deers(n):
    # Random deer for each of n images: returns (images [n, 3, 32, 32], bank indices [n])
    idx = torch.randint(0, K_DEER, (n,), device=device)
    return deer_bank[idx], idx


ncols = 10
fig, axes = plt.subplots((K_DEER + ncols - 1) // ncols, ncols, figsize=(1.6 * ncols, 1.8 * ((K_DEER + ncols - 1) // ncols)), squeeze=False)
for ax in axes.flat:
    ax.axis("off")
for k in range(K_DEER):
    ax = axes.flat[k]
    ax.imshow((deer_bank[k].cpu().permute(1, 2, 0) * 0.5 + 0.5).clamp(0, 1))
    ax.set_title(f"#{DEER_IDX[k]}", fontsize=8)
plt.suptitle("Target deer bank"); plt.tight_layout(); plt.show()
```

## Step 3: training loop

Details:
* **Differentiable augmentation** (random flip and random translation of up to 3 px) is applied to the poisoned image before the surrogates. This makes the perturbation robust to the crops and flips the victim/expert models train with, and improves transfer.
* **Straight-through 8-bit quantisation**: the poisoner saves images as PIL `uint8`, so the encoder is trained on quantised outputs (rounding in the forward pass, identity gradient).
* The loss is averaged over the 3 surrogates. Evaluation reports, on **test** trucks, the fraction each surrogate classifies as deer, plus the realised $L_\infty$ and PSNR.

```python
def quantize_st(x):
    # Straight-through 8-bit quantisation in [-1, 1] space
    x01 = (x + 1) / 2
    xq = torch.round(x01 * 255) / 255
    return (x01 + (xq - x01).detach()) * 2 - 1


def diff_augment(x, max_shift=3):
    # Random horizontal flip + random translation (reflect padding), differentiable w.r.t. x
    B = x.size(0)
    flip = torch.rand(B, device=x.device) < 0.5
    x = torch.where(flip[:, None, None, None], x.flip(3), x)
    pad = F.pad(x, [max_shift] * 4, mode="reflect")
    dx, dy = np.random.randint(0, 2 * max_shift + 1, size=2)
    return pad[:, :, dy:dy + 32, dx:dx + 32]


def surrogate_losses(x_p, deer_idx):
    # deer_idx[i] = index in deer_bank of the deer hidden in x_p[i]
    feat_loss = ce_loss = 0.0
    fooled = []
    y = torch.full((x_p.size(0),), deer_index, device=x_p.device, dtype=torch.long)
    for m, f_bank in zip(surrogates, deer_feats):
        f = resnet18_features(m, x_p)
        logits = m.linear(f)
        f_deer = f_bank[deer_idx]                                   # [B, 512]
        feat_loss = feat_loss + (((f - f_deer) ** 2).sum(1) / (f_deer ** 2).sum(1)).mean()
        ce_loss = ce_loss + F.cross_entropy(logits, y)
        fooled.append((logits.argmax(1) == deer_index).float().mean().item())
    K = len(surrogates)
    return feat_loss / K, ce_loss / K, fooled


@torch.no_grad()
def evaluate_deer_encoder(encoder, loader, disc=None):
    encoder.eval()
    hits = torch.zeros(len(surrogates)); n = 0; linf = 0.0; mse = 0.0; d_acc = 0.0
    for x, _ in loader:
        x = x.to(device)
        deers, _ = sample_deers(x.size(0))
        x_p = quantize_st(encoder(x, deers))
        if disc is not None:
            d_acc += discriminator_loss(disc, x, x_p)[1].item() * x.size(0)
        for k, m in enumerate(surrogates):
            hits[k] += (m(x_p).argmax(1) == deer_index).sum().item()
        linf = max(linf, (x_p - x).abs().max().item() / 2)             # back to [0, 1] scale
        mse += (((x_p - x) / 2) ** 2).mean(dim=(1, 2, 3)).sum().item()
        n += x.size(0)
    psnr = 10 * np.log10(1.0 / (mse / n))
    return (hits / n).tolist(), linf, psnr, (d_acc / n if disc is not None else None)


def train_deer_encoder(encoder, loader, epochs=30, lr=1e-3,
                       lam_feat=1.0, lam_ce=1.0, lam_l2=0.5, lam_tv=1.0, lam_sat=1.0,
                       disc=None, lam_adv=0.1, disc_lr=2e-4, adv_start_epoch=3, d_steps=1,
                       eval_every=5):
    # Imperceptibility terms are expressed in units of eps (L2 ~ (delta/eps)^2, TV ~ |grad delta|/eps),
    # so they are O(1) and comparable to the CE / feature losses.
    #
    # If `disc` is given, a clean-vs-poisoned discriminator is trained alternately with the encoder:
    #   D step: BCE(D(clean), 1) + BCE(D(poisoned.detach()), 0)
    #   G step: lam_adv * BCE(D(poisoned), 1)   (non-saturating), only from `adv_start_epoch` on,
    #           so the encoder first learns the attack and D gets a head start.
    opt = torch.optim.Adam(encoder.parameters(), lr=lr)
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs * len(loader))
    if disc is not None:
        opt_d = torch.optim.Adam(disc.parameters(), lr=disc_lr, betas=(0.5, 0.999))
    history = []
    for epoch in range(epochs):
        encoder.train()
        if disc is not None:
            disc.train()
        use_adv = disc is not None and epoch >= adv_start_epoch
        stats = np.zeros(9); n = 0
        for x, _ in loader:
            x = x.to(device)
            deers, deer_idx = sample_deers(x.size(0))
            delta, z = encoder.perturbation(x, deers, return_raw=True)
            x_p = quantize_st(torch.clamp(x + delta, -1, 1))

            # ---- discriminator step ----
            d_acc = float("nan")
            if disc is not None:
                for _ in range(d_steps):
                    d_loss, d_acc = discriminator_loss(disc, x, x_p.detach())
                    opt_d.zero_grad(set_to_none=True)
                    d_loss.backward()
                    opt_d.step()
                d_acc = d_acc.item()

            # ---- encoder step ----
            feat_loss, ce_loss, fooled = surrogate_losses(diff_augment(x_p), deer_idx)
            l2 = masked_l2_loss(delta, encoder.eps, x)     # heavier weight on flat regions
            tv = tv_loss(delta, encoder.eps)               # penalises high-frequency patterns
            sat = saturation_loss(z)                       # keeps tanh out of saturation (else no gradient)
            adv = generator_adv_loss(disc, x_p) if use_adv else torch.zeros((), device=device)
            loss = (lam_feat * feat_loss + lam_ce * ce_loss + lam_l2 * l2 + lam_tv * tv
                    + lam_sat * sat + lam_adv * adv)

            opt.zero_grad(set_to_none=True)
            loss.backward()                                # also leaves grads in disc; cleared at its next step
            opt.step(); sched.step()

            stats += np.array([loss.item(), feat_loss.item(), ce_loss.item(), l2.item(), tv.item(),
                               (delta.abs() / encoder.eps).mean().item(), np.mean(fooled),
                               adv.item(), d_acc]); n += 1
        stats /= n
        msg = (f"epoch {epoch+1}/{epochs} | loss {stats[0]:.3f} | feat {stats[1]:.3f} | CE {stats[2]:.3f} | "
               f"L2 {stats[3]:.3f} | TV {stats[4]:.3f} | mean|d|/eps {stats[5]:.3f} | train p(deer) {stats[6]:.3f}")
        if disc is not None:
            msg += f" | adv {stats[7]:.3f} | D acc {stats[8]:.3f}"
        if (epoch + 1) % eval_every == 0 or epoch == epochs - 1:
            rates, linf, psnr, d_test = evaluate_deer_encoder(encoder, truck_test_loader, disc)
            msg += f" || TEST deer rate {[round(r, 3) for r in rates]} | Linf {linf*255:.1f}/255 | PSNR {psnr:.1f} dB"
            if d_test is not None:
                msg += f" | D acc {d_test:.3f}"
            history.append(dict(epoch=epoch + 1, rates=rates, linf=linf, psnr=psnr, d_acc=d_test))
        print(msg)
    return history
```

```python
truck_train_loader = DataLoader(truck_dataset, batch_size=64, shuffle=True, num_workers=4, drop_last=True)

deer_encoder = DeerBlendEncoder(eps=EPS, mask_tau=MASK_TAU, mask_floor=MASK_FLOOR, blur_sigma=BLUR_SIGMA,
                                tanh_temp=TANH_TEMP, use_mask=False).to(device)
discriminator = PoisonDiscriminator().to(device)       # set to None to train without the adversarial term
history = train_deer_encoder(deer_encoder, truck_train_loader, epochs=40, lr=1e-3,
                             lam_feat=1.0, lam_ce=5.0, lam_l2=1.0, lam_tv=0.0, lam_sat=0.0,
                             disc=discriminator, lam_adv=50.0, adv_start_epoch=10)

torch.save({"state_dict": deer_encoder.state_dict(), "eps": deer_encoder.eps,
            "mask_tau": MASK_TAU, "mask_floor": MASK_FLOOR, "blur_sigma": BLUR_SIGMA, "tanh_temp": TANH_TEMP,
            "deer_idx": DEER_IDX, "deer_bank": deer_bank.cpu()},
           "noise_generator/deer_blend_encoder.pth")
```

### Post-hoc detectability check

The training discriminator co-evolves with the encoder, so "D acc ≈ 0.5" during training only means *that* D was fooled. A stronger test: freeze the encoder and train a **fresh** discriminator from scratch on train trucks, then measure its clean-vs-poisoned accuracy on test trucks. 0.5 = undetectable; 1.0 = trivially detectable.

```python
def fit_fresh_detector(encoder, epochs=10, lr=2e-4):
    encoder.eval()
    det = PoisonDiscriminator().to(device)
    opt_det = torch.optim.Adam(det.parameters(), lr=lr, betas=(0.5, 0.999))
    for epoch in range(epochs):
        det.train()
        for x, _ in truck_train_loader:
            x = x.to(device)
            with torch.no_grad():
                x_p = quantize_st(encoder(x, sample_deers(x.size(0))[0]))
            loss, _ = discriminator_loss(det, x, x_p)
            opt_det.zero_grad(set_to_none=True); loss.backward(); opt_det.step()
        _, _, _, acc = evaluate_deer_encoder(encoder, truck_test_loader, det)
        print(f"fresh detector epoch {epoch+1}/{epochs}: test clean-vs-poisoned accuracy {acc:.3f}")
    return det

fresh_detector = fit_fresh_detector(deer_encoder, epochs=10)
```

## Ablation: blur sigma x epsilon (no discriminator)

3 x 3 grid: `BLUR_SIGMA` in {0, 0.3, 0.5} and `EPS_PIXEL` in {2, 4, 6}/255. All other settings match the main run (cells 46 and 50), with `disc=None`. Each run is saved to `noise_generator/sweep_blur_eps/` (checkpoint + config + history); runs whose checkpoint already exists are loaded instead of retrained, so the cell can be resumed if interrupted.

```python
SWEEP_DIR = "noise_generator/sweep_blur_eps"
os.makedirs(SWEEP_DIR, exist_ok=True)

SWEEP_SIGMAS = [0.0, 0.3, 0.5]
SWEEP_EPS_PIXEL = [2 / 255, 4 / 255, 6 / 255]
SWEEP_EPOCHS = 40
SWEEP_USE_MASK = False          # same as the main run (cell 46)
SWEEP_LOSS = dict(lam_feat=1.0, lam_ce=5.0, lam_l2=1.0, lam_tv=0.0, lam_sat=0.0)


def sweep_name(sigma, eps_pixel):
    return f"sigma{sigma:.1f}_eps{round(eps_pixel * 255)}"


sweep_results = {}
for eps_pixel in SWEEP_EPS_PIXEL:
    for sigma in SWEEP_SIGMAS:
        name = sweep_name(sigma, eps_pixel)
        path = f"{SWEEP_DIR}/{name}.pth"
        if os.path.exists(path):
            print(f"=== {name}: loading {path}")
            sweep_results[name] = torch.load(path, map_location="cpu")
            continue

        print(f"\n=== {name}: training")
        torch.manual_seed(0)
        enc = DeerBlendEncoder(eps=2 * eps_pixel, mask_tau=MASK_TAU, mask_floor=MASK_FLOOR, blur_sigma=sigma,
                               tanh_temp=TANH_TEMP, use_mask=SWEEP_USE_MASK).to(device)
        hist = train_deer_encoder(enc, truck_train_loader, epochs=SWEEP_EPOCHS, lr=1e-3,
                                  disc=None, eval_every=10, **SWEEP_LOSS)
        result = {
            "state_dict": enc.state_dict(),
            "config": dict(eps_pixel=eps_pixel, eps=2 * eps_pixel, blur_sigma=sigma, mask_tau=MASK_TAU,
                           mask_floor=MASK_FLOOR, use_mask=SWEEP_USE_MASK, tanh_temp=TANH_TEMP,
                           epochs=SWEEP_EPOCHS, lr=1e-3, **SWEEP_LOSS),
            "history": hist,
            "deer_idx": DEER_IDX,
            "deer_bank": deer_bank.cpu(),
        }
        torch.save(result, path)
        sweep_results[name] = result

# Summary table (final evaluation of each run)
rows = []
for name, r in sweep_results.items():
    last = r["history"][-1]
    rows.append(dict(eps=f"{round(r['config']['eps_pixel'] * 255)}/255", sigma=r["config"]["blur_sigma"],
                     **{f"deer rate s{k}": round(v, 3) for k, v in enumerate(last["rates"])},
                     mean_deer_rate=round(float(np.mean(last["rates"])), 3),
                     linf=f"{last['linf'] * 255:.1f}/255", psnr=round(last["psnr"], 2)))
sweep_df = pd.DataFrame(rows).set_index(["eps", "sigma"])
sweep_df
```

```python
# Compare, for ONE truck, the perturbation produced by every (eps, sigma) run.
# Rows = epsilon, columns = [original | poisoned for each sigma | delta for each sigma]
TRUCK_SHOW = 0            # index into the test trucks
DEER_SHOW = 0             # index into deer_bank (same deer for every run)
COMMON_SCALE = True       # True: delta of every run scaled by the largest epsilon (magnitudes comparable)
                          # False: each delta scaled by its own epsilon (patterns easier to see)

x_one = truck_test_dataset[TRUCK_SHOW][0][None].to(device)
deer_one = deer_bank[DEER_SHOW][None]
to_img = lambda t: (t.cpu().permute(1, 2, 0) * 0.5 + 0.5).clamp(0, 1).numpy()
max_eps_pixel = max(SWEEP_EPS_PIXEL)

n_rows, n_sig = len(SWEEP_EPS_PIXEL), len(SWEEP_SIGMAS)
fig, axes = plt.subplots(n_rows, 1 + 2 * n_sig, figsize=(2.3 * (1 + 2 * n_sig), 2.5 * n_rows), squeeze=False)
for ax in axes.flat:
    ax.axis("off")

for r_i, eps_pixel in enumerate(SWEEP_EPS_PIXEL):
    axes[r_i, 0].imshow(to_img(x_one[0]), interpolation="nearest")
    axes[r_i, 0].set_title(f"original\n(eps {round(eps_pixel * 255)}/255 row)", fontsize=9)
    for c_i, sigma in enumerate(SWEEP_SIGMAS):
        r = sweep_results[sweep_name(sigma, eps_pixel)]
        cfg = r["config"]
        enc = DeerBlendEncoder(eps=cfg["eps"], mask_tau=cfg["mask_tau"], mask_floor=cfg["mask_floor"],
                               blur_sigma=cfg["blur_sigma"], tanh_temp=cfg["tanh_temp"],
                               use_mask=cfg["use_mask"]).to(device)
        enc.load_state_dict(r["state_dict"]); enc.eval()
        with torch.no_grad():
            x_p = quantize_st(enc(x_one, deer_one))
            preds = [dataset.classes[m(x_p).argmax(1).item()][:4] for m in surrogates]

        d = ((x_p - x_one) / 2)[0]                                            # [0, 1] scale
        scale = max_eps_pixel if COMMON_SCALE else cfg["eps_pixel"]
        psnr = 10 * np.log10(1.0 / d.pow(2).mean().item())
        deer_rate = np.mean(r["history"][-1]["rates"])

        ax = axes[r_i, 1 + c_i]
        ax.imshow(to_img(x_p[0]), interpolation="nearest")
        ax.set_title(f"poisoned, sigma {sigma}\n{'/'.join(preds)}", fontsize=9)

        ax = axes[r_i, 1 + n_sig + c_i]
        ax.imshow(((d / scale) * 0.5 + 0.5).cpu().permute(1, 2, 0).clamp(0, 1).numpy(), interpolation="nearest")
        ax.set_title(f"delta, sigma {sigma}\nPSNR {psnr:.1f} dB | rate {deer_rate:.2f}", fontsize=9)

scale_txt = f"all deltas scaled by {round(max_eps_pixel * 255)}/255" if COMMON_SCALE else "each delta scaled by its own eps"
fig.suptitle(f"Test truck #{TRUCK_SHOW}, deer #{DEER_IDX[DEER_SHOW]}  |  rows: eps = "
             f"{', '.join(f'{round(e * 255)}/255' for e in SWEEP_EPS_PIXEL)}  |  {scale_txt}", fontsize=11)
plt.tight_layout(); plt.show()
```

## Sweep: adversarial weight $\lambda_{\text{adv}}$ (fixed epsilon and sigma)

With the surrogate terms kept on, sweep only `lam_adv` to find the best trade-off between fooling the surrogates (deer rate) and evading detection. Epsilon is fixed at 6/255 and `BLUR_SIGMA` at `ADV_SWEEP_SIGMA` (0 by default, so the adversarial term is the only thing fighting detectability). `lam_adv = 0` is the no-discriminator baseline.

Detectability is read from a **freshly trained** `PoisonDiscriminator` (honest), not the co-adapted in-loop one. Uses `fit_fresh_detector` and `train_deer_encoder`, so run the earlier cells first. Independent of the epsilon x sigma sweep above (separate `adv_*` variables and `noise_generator/sweep_lambda_adv/`).

```python
ADV_SWEEP_DIR = "noise_generator/sweep_lambda_adv"
os.makedirs(ADV_SWEEP_DIR, exist_ok=True)

ADV_SWEEP_EPS_PIXEL = 6 / 255
ADV_SWEEP_SIGMA = 0.3
ADV_LAMBDAS = [1.0, 10.0, 20.0, 40.0, 60.0]   # 0.0 = no discriminator (baseline)
ADV_SWEEP_START = 10                                   # epoch at which the adversarial term switches on
ADV_SWEEP_EPOCHS = 40
ADV_SWEEP_USE_MASK = False
ADV_SWEEP_LOSS = dict(lam_feat=1.0, lam_ce=5.0, lam_l2=1.0, lam_tv=0.0, lam_sat=0.0)
ADV_FRESH_DET_EPOCHS = 8


def adv_name(adv):
    return f"adv{adv:g}"


adv_results = {}
for adv in ADV_LAMBDAS:
    name = adv_name(adv)
    path = f"{ADV_SWEEP_DIR}/{name}.pth"
    if os.path.exists(path):
        print(f"=== {name}: loading {path}")
        adv_results[name] = torch.load(path, map_location="cpu")
        continue

    print(f"\n=== {name}: training (lam_adv={adv}, sigma={ADV_SWEEP_SIGMA}, eps={round(ADV_SWEEP_EPS_PIXEL * 255)}/255)")
    torch.manual_seed(0)
    enc = DeerBlendEncoder(eps=2 * ADV_SWEEP_EPS_PIXEL, mask_tau=MASK_TAU, mask_floor=MASK_FLOOR,
                           blur_sigma=ADV_SWEEP_SIGMA, tanh_temp=TANH_TEMP, use_mask=ADV_SWEEP_USE_MASK).to(device)
    disc = PoisonDiscriminator().to(device) if adv > 0 else None       # adv == 0: pure baseline, no discriminator
    hist = train_deer_encoder(enc, truck_train_loader, epochs=ADV_SWEEP_EPOCHS, lr=1e-3,
                              disc=disc, lam_adv=adv, adv_start_epoch=ADV_SWEEP_START,
                              eval_every=10, **ADV_SWEEP_LOSS)

    det = fit_fresh_detector(enc, epochs=ADV_FRESH_DET_EPOCHS)
    _, _, _, fresh_acc = evaluate_deer_encoder(enc, truck_test_loader, det)
    print(f"    fresh-detector accuracy: {fresh_acc:.3f}")

    result = {
        "state_dict": enc.state_dict(),
        "config": dict(eps_pixel=ADV_SWEEP_EPS_PIXEL, eps=2 * ADV_SWEEP_EPS_PIXEL, blur_sigma=ADV_SWEEP_SIGMA,
                       lam_adv=adv, adv_start_epoch=ADV_SWEEP_START, mask_tau=MASK_TAU, mask_floor=MASK_FLOOR,
                       use_mask=ADV_SWEEP_USE_MASK, tanh_temp=TANH_TEMP, epochs=ADV_SWEEP_EPOCHS, lr=1e-3, **ADV_SWEEP_LOSS),
        "history": hist,
        "fresh_detector_acc": fresh_acc,
        "deer_idx": DEER_IDX,
        "deer_bank": deer_bank.cpu(),
    }
    torch.save(result, path)
    adv_results[name] = result

# Summary: attack strength (deer rate) vs detectability (fresh detector) vs quality (PSNR)
rows = []
for name, r in adv_results.items():
    last = r["history"][-1]; cfg = r["config"]
    rows.append(dict(lam_adv=cfg["lam_adv"],
                     mean_deer_rate=round(float(np.mean(last["rates"])), 3),
                     train_D_acc=(round(last["d_acc"], 3) if last.get("d_acc") is not None else None),
                     fresh_D_acc=round(r["fresh_detector_acc"], 3),
                     psnr=round(last["psnr"], 2)))
adv_df = pd.DataFrame(rows).set_index("lam_adv").sort_index()
adv_df
```

```python
# Visualise the lambda_adv tuning: how each metric moves with lam_adv, and the
# attack-strength (deer rate) vs detectability (fresh detector) trade-off.
# Falls back to reading ADV_SWEEP_DIR if adv_results isn't in memory (e.g. after a kernel restart).
if "adv_results" not in globals():
    adv_results = {f.replace(".pth", ""): torch.load(f"{ADV_SWEEP_DIR}/{f}", map_location="cpu")
                   for f in sorted(os.listdir(ADV_SWEEP_DIR)) if f.endswith(".pth")}

runs = sorted(adv_results.values(), key=lambda r: r["config"]["lam_adv"])
lam = [r["config"]["lam_adv"] for r in runs]
deer = [float(np.mean(r["history"][-1]["rates"])) for r in runs]
fresh = [r["fresh_detector_acc"] for r in runs]
train_d = [r["history"][-1].get("d_acc") for r in runs]
psnr = [r["history"][-1]["psnr"] for r in runs]
xpos = range(len(lam))                                   # even spacing (lam_adv includes 0)
labels = [f"{l:g}" for l in lam]

fig, (axL, axR) = plt.subplots(1, 2, figsize=(14, 5.2))

# --- left: metrics vs lam_adv ---
axL.plot(xpos, deer, "o-", color="tab:green", label="deer rate (attack)")
axL.plot(xpos, fresh, "s-", color="tab:red", label="fresh detector acc (detectability)")
td_x = [i for i, t in zip(xpos, train_d) if t is not None]
td_y = [t for t in train_d if t is not None]
axL.plot(td_x, td_y, "^--", color="tab:orange", alpha=0.7, label="in-loop D acc (co-adapted)")
axL.axhline(0.5, color="grey", ls=":", lw=1, label="detector chance (0.5)")
axL.set_xticks(list(xpos)); axL.set_xticklabels(labels)
axL.set_xlabel(r"$\lambda_{\mathrm{adv}}$"); axL.set_ylabel("rate / accuracy")
axL.set_ylim(0, 1.02); axL.set_title(f"Metrics vs $\\lambda_{{\\mathrm{{adv}}}}$  (eps {round(runs[0]['config']['eps_pixel']*255)}/255, sigma {runs[0]['config']['blur_sigma']})")
axL.legend(loc="center right", fontsize=9); axL.grid(alpha=0.3)
axP = axL.twinx()
axP.plot(xpos, psnr, "d:", color="tab:blue", alpha=0.6)
axP.set_ylabel("PSNR (dB)", color="tab:blue"); axP.tick_params(axis="y", labelcolor="tab:blue")

# --- right: trade-off, attack (y) vs detectability (x) ---
sc = axR.scatter(fresh, deer, c=xpos, cmap="viridis", s=90, zorder=3)
for f, d, lb in zip(fresh, deer, labels):
    axR.annotate(f"$\\lambda$={lb}", (f, d), textcoords="offset points", xytext=(7, 4), fontsize=9)
axR.set_xlabel("fresh detector accuracy  (lower = harder to detect ->)")
axR.set_ylabel("deer rate  (higher = stronger attack ^)")
axR.set_title("Trade-off: attack strength vs detectability")
axR.annotate("better", xy=(0.52, max(deer)), xytext=(0.72, max(deer) * 0.62),
             arrowprops=dict(arrowstyle="->", color="grey"), color="grey", fontsize=10)
axR.grid(alpha=0.3)

plt.tight_layout(); plt.show()

# Knee: best deer rate among runs whose fresh detector is within 0.02 of the lowest achieved
best_det = min(fresh)
cand = [(l, d, f) for l, d, f in zip(lam, deer, fresh) if f <= best_det + 0.02]
knee = max(cand, key=lambda t: t[1])
print(f"lowest fresh-detector acc: {best_det:.3f}")
print(f"best attack near that detectability: lam_adv={knee[0]:g}  ->  deer rate {knee[1]:.3f}, detector {knee[2]:.3f}")
```
